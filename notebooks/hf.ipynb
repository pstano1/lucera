```python
import duckdb
import pandas as pd
from datasets import Dataset
import sys
import importlib
sys.path.insert(0, '/Users/pstano/Documents/applications/lucera')

import utilities
importlib.reload(utilities)

from utilities.region_manager import RegionIndex
from pathlib import Path

BASE_DIR = Path('/Users/pstano/Documents/applications/lucera')

idx = RegionIndex(path=str(BASE_DIR / 'data/index.parquet'))

COUNTRY_CONFIGS = {
    'PT': ('europe', 'pt'),
    'IS': ('europe', 'is'),
    'GNB': ('africa', 'gnb'),
}

for country, (continent, code) in COUNTRY_CONFIGS.items():
    complete = idx.complete()
    complete = complete[complete['country_code'] == country]
    complete = complete[complete['region_id'] != 'NA']
    
    if len(complete) == 0:
        print(f"No complete regions for {country}")
        continue
    
    admin_groups = complete['output_path'].apply(
        lambda p: p.split('/')[4]
    ).unique()
    
    for admin in admin_groups:
        admin_regions = complete[complete['output_path'].str.contains(f'/{code}/{admin}/')]
        paths = admin_regions['output_path'].tolist()
        existing = [str(BASE_DIR / p) for p in paths if (BASE_DIR / p).exists()]
        
        if not existing:
            continue
        
        paths_str = ', '.join([f"'{p}'" for p in existing])
        df = duckdb.query(f"SELECT * FROM read_parquet([{paths_str}])").df()
        print(f"{country}/{admin}: {len(df)} points")
        
        ds = Dataset.from_pandas(df)
        ds.push_to_hub(
            "pstano/lucera",
            config_name=f"{admin}_{country.lower()}_r7_2023"
        )
        print(f"Pushed {country}/{admin}")
```
